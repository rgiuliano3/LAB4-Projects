```python
## Second cell
# run this for step 19 in your instructions

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Part 3: Accessing your data
# edit this code to load your data into your workspace

data = pd.read_csv("/workspaces/LAB4-Projects/data/raw/imdb_movies.csv") #add your data file path to this line!
data.head()
```

```python
## Checking dimensions

data.shape
```

```python
## Column information

data.info()
```

```python
## Data statistics

data.describe()
```

```python
## Scatterplot

fig, ax = plt.subplots(figsize=(15, 10))          # blank canvas, 15 x 10 inches
ax.scatter(data['budget_x'], data['revenue'])           # x = budget, y = revenue

plt.xlabel("Budget")                     # label the x-axis
plt.ylabel("Revenue")     # label the y-axis
plt.title("Revenue vs. Budget")  # give it a title
plt.show()                                        # display the chart
```

```python
## Bar Chart

country_counts = data['country'].value_counts()
country_counts

fig, ax = plt.subplots(figsize=(15, 10))
ax.bar(country_counts.index, country_counts.values)   # x = category names, height = counts
ax.tick_params("x", rotation=45)                    # tilt the x-axis labels 45 degrees

plt.ylabel("count")
plt.xlabel("country")
plt.title("Number of movies per country")
plt.show()
```

```python
## Density Plot

plt.figure(figsize=(15, 10))
sns.kdeplot(data=data, x="score")

plt.title("Distribution of movie scores")
plt.xlabel("Movie scores")
plt.show()
```

```python
## Correlation Matrix

correlation_matrix = data[['budget_x', 'revenue', 'score']].corr()
correlation_matrix
```
